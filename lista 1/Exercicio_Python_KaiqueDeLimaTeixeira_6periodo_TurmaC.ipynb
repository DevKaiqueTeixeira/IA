```python
import math

print("Exercício 3 cálculo da área do círuclo")

raio = float(input("Digite o raio: "))

area = math.pi * raio ** 2

print(f"A área do círculo é: {area:.2f}")
```

```python
print("Exercício 7 número impar ou par")

numero = float(input("Digite um número"))

if numero % 2 == 1:
  print(f"O número {numero} é ímpar")
else:
  print(f"O número {numero} é par")
```

```python
print("Exerício 9 Fatorial com lib")

import math

numero = int(input("Digite um número: "))
print(math.factorial(numero))

print("Exerício 9 Fatorial com sem lib")

fatorial = int(1)

for i in range(1, numero + 1):
  fatorial *= i

print(f"Resultado fatorial sem lib: {fatorial}")
```

```python
print("Exercício 17 Dicionário de Contagem de Palavras")

texto = input("Digite uma frase: ")

palavras = texto.split()

contagem = {}

for palavra in palavras:
    if palavra in contagem:
        contagem[palavra] += [1]
    else:
        contagem[palavra] = 1

print(f"Quantidade de vezes repetidas: {contagem}")
```
