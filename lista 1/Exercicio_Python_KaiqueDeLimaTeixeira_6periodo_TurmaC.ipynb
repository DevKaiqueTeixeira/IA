```python
print("Exercício 17 Dicionário de Contagem de Palavras")

texto = input("Digite uma frase: ")

palavras = texto.split()

contagem = {}

for palavra in palavras:
    if palavra in contagem:
        contagem[palavra] += [1]
    else:
        contagem[palavra] = 1

print(f"Quantidade de vezes repetidas: {contagem}")
```

```python
print("Exercício 21 Lista Ordenada binária")

lista = [10, 20, 30, 40, 50, 60, 70, 80, 90]
numero = 70

inicio = 0
fim = len(lista) - 1
#print(len(lista) - 1)
while inicio <= fim:

    meio = (inicio + fim) // 2
    print(meio)

    if lista[meio] == numero:
        print("Elemento encontrado na posição:", meio)
        break

    elif lista[meio] < numero:
        inicio = meio + 1
        print("Numero é maior")

    else:
        fim = meio - 1
        print("Numero é maior")
else:
    print("Elemento não encontrado")
```

```python
print("Exercício 22 - Cifra de César")
def cifra_cesar_reverso(mensagem, deslocamento):
    resultado = ""

    for letra in mensagem:
        letra = letra.lower()

        if letra.isalpha():

            # Converte a letra para número
            valor_letra = ord(letra)

            # Converte para uma posição de 0 a 25
            posicao = valor_letra - ord('a')

            # Adiciona o deslocamento
            nova_posicao = posicao - deslocamento

            # Faz voltar para o início quando passar do Z
            nova_posicao = nova_posicao % 26

            # Volta a posição para o código da letra
            novo_valor = nova_posicao + ord('a')

            # Converte o número novamente para uma letra
            nova_letra = chr(novo_valor)

            # Adiciona a letra ao resultado
            resultado += nova_letra

        else:
            resultado += letra

    return resultado


def cifra_cesar(mensagem, deslocamento):
    resultado = ""

    for letra in mensagem:
        letra = letra.lower()

        if letra.isalpha():

            # Converte a letra para número
            valor_letra = ord(letra)

            # Converte para uma posição de 0 a 25
            posicao = valor_letra - ord('a')

            # Adiciona o deslocamento
            nova_posicao = posicao + deslocamento

            # Faz voltar para o início quando passar do Z
            nova_posicao = nova_posicao % 26

            # Volta a posição para o código da letra
            novo_valor = nova_posicao + ord('a')

            # Converte o número novamente para uma letra
            nova_letra = chr(novo_valor)

            # Adiciona a letra ao resultado
            resultado += nova_letra

        else:
            resultado += letra

    return resultado


mensagem = input("Digite a mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

mensagem_criptografada = cifra_cesar(mensagem, deslocamento)

mensagem_criptografada_reverso = cifra_cesar_reverso(mensagem_criptografada, deslocamento)

print("Mensagem criptografada:", mensagem_criptografada)

print("Mensagem Original:", mensagem_criptografada_reverso)

```
