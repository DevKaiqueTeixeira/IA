```python
print("Exercício 21 Lista Ordenada binária")

lista = [10, 20, 30, 40, 50, 60, 70, 80, 90]
numero = 70

inicio = 0
fim = len(lista) - 1
#print(len(lista) - 1)
while inicio <= fim:

    meio = (inicio + fim) // 2
    print(meio)

    if lista[meio] == numero:
        print("Elemento encontrado na posição:", meio)
        break

    elif lista[meio] < numero:
        inicio = meio + 1
        print("Numero é maior")

    else:
        fim = meio - 1
        print("Numero é maior")
else:
    print("Elemento não encontrado")
```

```python
print("Exercício 27 Anagramas")

palavra1 = input("Digite a primeira palavra: ")
lista1=[]
palavra2 = input("Digite a primeira palavra: ")
lista2=[]

for i in palavra1:
  lista1.append(i)

for i in palavra2:
  lista2.append(i)

for i in lista1:

  if(i in lista2):
    resultado = True
    lista2.remove(i)
  else:
    resultado = False
    break;

if(resultado):
  print("Palavras iguais")
else:
  print("Palavras diferentes")
```

```python
print("Exercício 28 Gráfico de Barras")

import matplotlib.pyplot as plt

categorias = []
valores = []

quantidade = int(input("Quantos dados você deseja inserir? "))

for i in range(quantidade):
    categoria = input(f"Digite o nome do {i + 1}º dado: ")
    valor = int(input(f"Digite o valor de {categoria}: "))

    categorias.append(categoria)
    valores.append(valor)

plt.bar(categorias, valores)

plt.title("Gráfico de Barras")
plt.xlabel("Categorias")
plt.ylabel("Valores")

plt.show()
```

```python
print("Exercício 29 FizzBuzz")

listaFB=[]
listaF=[]
listaB=[]
listaNenhum=[]

for i in range(1, 101):

    if i % 3 == 0 and i % 5 == 0:
        listaFB.append(i)

    elif i % 3 == 0:
        listaF.append(i)

    elif i % 5 == 0:
        listaB.append(i)

    else:
      listaNenhum.append(i)


print(f"FizzBuzz: {listaFB}, Fizz: {listaF}, Buzz: {listaB}, Não divisivel por nenhum: {listaNenhum}")
```

```python
print("Exercício 30 - Jogo da Forca")

texto = input("Digite a palavra da Forca (não mostre pra ninguém): ")
dica = input("Digite uma dica: ")

lista = []
chances = 3
resultado = []

for i in texto:
    lista.append(i)
    resultado.append("_")

while True:

    print(" ".join(resultado))

    opcao = input(f"Digite uma letra a dica é: {dica} você tem {chances} chances ")

    if opcao in lista:

        for i in range(len(lista)):
            if opcao == lista[i] and chances == 0:
                resultado[i] = opcao

        print("Acertou!")

        if resultado == lista:
            print("Você ganhou!")
            break

    else:
        print("Errou!")
        chances = chances - 1
        if(chances == 0):
          print("Tente novamente, suas chances acabaram ")
          break
```
