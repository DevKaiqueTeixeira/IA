```python
import math

print("Exercício 3 cálculo da área do círuclo")

raio = float(input("Digite o raio: "))

area = math.pi * raio ** 2

print(f"A área do círculo é: {area:.2f}")
```

```python
print("Exercício 4 Conversão temperatura")

temperatura = float(input("Forneça uma temperatura em Celsius: "))

fahrenheit = temperatura * (9/5) + 32

print(f"Temperatura em Fahrenheit: {fahrenheit:.2f} °F")
```

```python
print("Exercício 5 Tabuada")

numero = float(input("Forneça o número para a tabuada: "))
for i in range(0, 11):
    resultado = numero * i
    print(f"{numero} x {i} = {resultado}")
```

```python
print("Exercício 6 Números pares de 1 a 20")

for i in range(0,21):
    if i % 2 == 0:
        print(i)
```

```python
print("Exercício 7 número impar ou par")

numero = float(input("Digite um número"))

if numero % 2 == 1:
  print(f"O número {numero} é ímpar")
else:
  print(f"O número {numero} é par")
```

```python
print("Exercício 8 Palíndromo")

palavra = (input("Digite uma palavra: "))

if palavra == palavra[::-1]:
    print("É um palíndromo")
else:
    print("Não é um palíndromo")
```

```python
print("Exerício 9 Fatorial com lib")

import math

numero = int(input("Digite um número: "))
print(math.factorial(numero))

print("Exerício 9 Fatorial com sem lib")

fatorial = int(1)

for i in range(1, numero + 1):
  fatorial *= i

print(f"Resultado fatorial sem lib: {fatorial}")
```
