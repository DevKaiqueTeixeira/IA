```python
print("Exercício  Manipulação array")
import numpy as np

#array 3x3
array = np.random.randint(0, 10, (3, 3))
print(f"Array padrão: ")
print(array)

max_value =  np.max(array)
print(f"Maior número: {max_value}")

min_value =  np.min(array)
print(f"Maior número: {min_value}")

sum_value = np.sum(array)
print(f"Soma dos elementos: {sum_value}")

```

```python
print("Exercício  Operações matemática ")

import numpy as np

array = np.array([[1,2], [3,4]])

array2 = np.array([[1,2], [3,4]])

soma = array + array2

print(f"Soma dos arrays: {soma}")

multiplicar = array * array2
print(f"Multiplicar dos arrays: {multiplicar}")

subtrair = array - array2
print(f"Subtrair dos arrays: {subtrair}")



```
