```python
print("Exercicio 2 scipy")

import numpy as np
from scipy.integrate import solve_ivp

def f(x, y):
    return x - y

solucao = solve_ivp(
    f,
    [0, 10],
    [1],
    t_eval=np.linspace(0, 10, 100)
)

print("Valores de x:")
print(solucao.t)

print("Valores de y:")
print(solucao.y[0])
```

```python
print("Exercicio 3 scipy")

from scipy.optimize import root

def f(x):
    return x**4 - 3*x**3 + 2

chutes = [-2, 0, 1, 2, 3, 4]

raizes = []

for chute in chutes:
    resultado = root(f, chute)

    if resultado.success:
        raiz = resultado.x[0]

        if not any(abs(raiz - r) < 0.0001 for r in raizes):
            raizes.append(raiz)

print("Raízes:", raizes)
```

```python
print("Exercicio 5 scipy")

import numpy as np
from scipy.fft import fft, fftfreq

t = np.linspace(0, 1, 1000)

sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

transformada = fft(sinal)

frequencias = fftfreq(len(t), t[1] - t[0])

print("Transformada de Fourier:")
print(transformada)

print("Frequências:")
print(frequencias)
```

```python
print("Exercicio 20 - Pandas")

import pandas as pd

df = pd.DataFrame({
    "Categoria": ["A", "A", "B", "B", "C", "C"],
    "Valor": [10, 20, 15, 25, 30, 40]
})

resultado = df.groupby("Categoria")["Valor"].median()

print(resultado)
```

```python
print("Exercicio 6 - Matplotlib")

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.1, 10, 100)

y1 = np.exp(x)
y2 = np.log(x)

plt.subplot(2, 1, 1)
plt.plot(x, y1)
plt.title("y = eˣ")

plt.subplot(2, 1, 2)
plt.plot(x, y2)
plt.title("y = log(x)")

plt.tight_layout()

plt.show()
```

```python
print("Exercicio 8 - Matplotlib")

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

erro = 0.2

plt.plot(x, y)

plt.fill_between(
    x,
    y - erro,
    y + erro,
    alpha=0.3
)

plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Função y = sin(x) com Intervalo de Confiança")

plt.show()
```

```python
print("Exercicio 11 - Matplotlib")

import matplotlib.pyplot as plt

categorias = ["A", "B", "C"]

grupo1 = [10, 20, 15]
grupo2 = [15, 10, 20]
grupo3 = [5, 15, 10]

plt.bar(categorias, grupo1, label="Grupo 1")
plt.bar(categorias, grupo2, bottom=grupo1, label="Grupo 2")

base_grupo3 = [
    grupo1[i] + grupo2[i]
    for i in range(3)
]

plt.bar(
    categorias,
    grupo3,
    bottom=base_grupo3,
    label="Grupo 3"
)

plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.title("Gráfico de Barras Empilhadas")
plt.legend()

plt.show()
```
