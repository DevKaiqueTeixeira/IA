```python
print("Exercicio 2 scipy")

import numpy as np
from scipy.integrate import solve_ivp

def f(x, y):
    return x - y

solucao = solve_ivp(
    f,
    [0, 10],
    [1],
    t_eval=np.linspace(0, 10, 100)
)

print("Valores de x:")
print(solucao.t)

print("Valores de y:")
print(solucao.y[0])
```

```python
print("Exercicio 3 scipy")

from scipy.optimize import root

def f(x):
    return x**4 - 3*x**3 + 2

chutes = [-2, 0, 1, 2, 3, 4]

raizes = []

for chute in chutes:
    resultado = root(f, chute)

    if resultado.success:
        raiz = resultado.x[0]

        if not any(abs(raiz - r) < 0.0001 for r in raizes):
            raizes.append(raiz)

print("Raízes:", raizes)
```

```python
print("Exercicio 5 scipy")

import numpy as np
from scipy.fft import fft, fftfreq

t = np.linspace(0, 1, 1000)

sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

transformada = fft(sinal)

frequencias = fftfreq(len(t), t[1] - t[0])

print("Transformada de Fourier:")
print(transformada)

print("Frequências:")
print(frequencias)
```

```python
print("Exercicio 9 scipy")

import numpy as np
from scipy.optimize import curve_fit

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

def funcao(x, a, b, c):
    return a*x**2 + b*x + c

parametros, _ = curve_fit(funcao, x, y)

print("Coeficientes:")
print("a =", parametros[0])
print("b =", parametros[1])
print("c =", parametros[2])
```
